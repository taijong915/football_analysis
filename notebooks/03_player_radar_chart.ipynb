# 🍕 선수 프로필 비교: Pizza & Radar Charts

`mplsoccer.PyPizza`를 활용하여 선수의 세부 스탯 백분위수(Percentile)를 직관적인 피자 차트/레이더 차트로 시각화합니다.

```python
import matplotlib.pyplot as plt
from mplsoccer import PyPizza

# 분석 지표 파라미터 (공격 3개, 패스/전개 3개, 수비/활동량 2개)
params = [
    "Non-Penalty Goals",
    "xG (Expected Goals)",
    "Shots Total",
    "Key Passes",
    "Passes into Box",
    "Progressive Passes",
    "Dribbles Completed",
    "Tackles & Interceptions"
]

# 백분위수 값 (0 ~ 100)
values = [94, 89, 93, 98, 95, 96, 91, 52]

# 카테고리별 색상 구분 (공격: 블루, 패스/빌드업: 오렌지, 수비: 레드)
slice_colors = ["#1A78CF"] * 3 + ["#FF9300"] * 3 + ["#D70232"] * 2
text_colors = ["#F2F2F2"] * len(params)

baker = PyPizza(
    params=params,
    background_color="#131313",
    straight_line_color="#EBEBE9",
    straight_line_lw=1,
    last_circle_lw=1,
    other_circle_lw=1,
    other_circle_ls="-."
)

fig, ax = baker.make_pizza(
    values,
    figsize=(8, 8),
    param_location=110,
    slice_colors=slice_colors,
    value_colors=text_colors,
    value_bck_colors=slice_colors,
    kwargs_slices=dict(edgecolor="#F2F2F2", zorder=2, linewidth=1),
    kwargs_params=dict(color="#F2F2F2", fontsize=11, va="center"),
    kwargs_values=dict(color="#F2F2F2", fontsize=10, zorder=3,
                       bbox=dict(edgecolor="#F2F2F2", facecolor="1", boxstyle="round,pad=0.2", lw=1))
)

fig.text(0.515, 0.97, "Lionel Messi - Attacking Midfielder / Winger", size=17, ha="center", color="#F2F2F2", weight="bold")
fig.text(0.515, 0.93, "Percentile Rank vs Top 5 European Leagues & International Competitions", size=10, ha="center", color="#B0B0B0")

plt.show()
```
